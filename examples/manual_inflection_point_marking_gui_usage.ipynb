## 3. Expected input files

By default, the GUI looks under `GT_GUI_Parsing/` for these files:

- `Upsamp_UP_Dict.pkl`
- `hs_manually_parsed_data.pkl`
- `to_manually_parsed_data.pkl`
- `marking_time.npy`

If `Upsamp_UP_Dict.pkl` is not found, the GUI opens a file picker so you can choose a source file manually.

## 4. Basic workflow inside the GUI

1. Ensure that the input data is a .pkl file containing a dictionary with format: key: dataset_name, value: numpy array of data.
2. Select the .pkl file to load into the GUI from the interactive window.
3. Click near desired signal point to mark, such as a heel-strike or toe-off if parsing underfoot pressure data.
4. If you want to remove a point, click again near the existing marker to remove it.
5. Use the toggle button in the bottom right to switch between Heel Strike and Toe Off labeling.
6. Use the forward/back buttons to move between datasets, progress is saved when moving between datasets.
7. Closing the window will autosave all progress.
8. hs_manually_parsed_data.pkl and to_manually_parsed_data.pkl contain saved data markers.
9. To ensure that saved data from the .pkl files loads again, ensure that the .pkl save files are within the same folder the were saved into from the GUI.

## 5. Platform-specific Qt binding behavior

The packaged GUI now selects its Qt binding automatically:

- macOS: `PyQt5`
- Windows: `PySide6`
- other platforms: `PyQt5` fallback

Make sure the matching dependency is installed in your environment.
