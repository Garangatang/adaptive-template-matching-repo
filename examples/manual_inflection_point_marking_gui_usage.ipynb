## 3. Expected input files

By default, the GUI looks under `GT_GUI_Parsing/` for these files:

- `Upsamp_UP_Dict.pkl`
- `hs_manually_parsed_data.pkl`
- `to_manually_parsed_data.pkl`
- `marking_time.npy`

If `Upsamp_UP_Dict.pkl` is not found, the GUI opens a file picker so you can choose a source file manually.

## 4. Basic workflow inside the GUI

1. Open or auto-load the source pressure dictionary.
2. Review the plotted signal for the current dataset.
3. Click near heel-strike or toe-off points to add markers.
4. Click again near an existing marker to remove it.
5. Use the toggle button to switch between Heel Strike and Toe Off labeling.
6. Use the forward/back buttons to move across datasets.
7. Save and close the window to persist labels and elapsed marking time.

## 5. Platform-specific Qt binding behavior

The packaged GUI now selects its Qt binding automatically:

- macOS: `PyQt5`
- Windows: `PySide6`
- other platforms: `PyQt5` fallback

Make sure the matching dependency is installed in your environment.

## 6. Notes for notebook environments

- GUI launch works best from a normal desktop Python session or terminal.
- If you are running inside Jupyter, you may prefer launching the GUI from a terminal tab rather than from a notebook kernel.
- The GUI writes label outputs back to pickle and NumPy files when the window closes.
